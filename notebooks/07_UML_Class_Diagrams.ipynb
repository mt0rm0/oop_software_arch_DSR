## 1. Why Sketch First? <a id="1"></a>

A 5-minute diagram can save hours of refactoring. Before writing code:
- You spot missing relationships
- You find responsibilities assigned to the wrong class
- You can discuss design with a colleague without opening an IDE

UML class diagrams are the standard notation. We'll use **Mermaid** which renders in VS Code and GitHub.

> To render in VS Code: install the **Markdown Preview Mermaid Support** extension  
> To render in PyCharm: Settings → Languages & Frameworks → Markdown → Enable Mermaid

## 2. The Essentials: Classes, Attributes, Methods <a id="2"></a>

Each box has three sections: class name, attributes (with types), methods (with return types).

Visibility prefix: `+` public, `-` private, `#` protected

```mermaid
classDiagram
    class Dataset {
        +DataFrame X
        +Series y
        +str name
        -dict _splits
        +split(test_size, random_state) Dataset
        +summary() str
        +train() tuple
        +test() tuple
    }

    class BaseTransformer {
        <<abstract>>
        +fit(X)* BaseTransformer
        +transform(X)* DataFrame
        +fit_transform(X) DataFrame
    }

    class StandardScaler {
        -Series _mean
        -Series _std
        +fit(X) StandardScaler
        +transform(X) DataFrame
    }

    class MinMaxScaler {
        -Series _min
        -Series _range
        +fit(X) MinMaxScaler
        +transform(X) DataFrame
    }

    BaseTransformer <|-- StandardScaler : inherits
    BaseTransformer <|-- MinMaxScaler : inherits
```

The `<<abstract>>` tag marks the ABC. Methods with `*` are abstract (must be overridden).

## 3. Relationships: Inheritance, Composition, Association <a id="3"></a>

| Arrow | Meaning | Python |
|---|---|---|
| `A <\|-- B` | B inherits from A | `class B(A)` |
| `A *-- B` | A is composed of B (B can't exist without A) | `self.b = B()` in `__init__` |
| `A o-- B` | A has a B (B exists independently) | method parameter or optional attribute |
| `A --> B` | A uses B | function parameter |

Composition is the most important for ML pipelines — a `Pipeline` *owns* its transformers.

```mermaid
classDiagram
    class MLPipeline {
        +str name
        -list _transformers
        -BaseModel _model
        +add_transformer(t BaseTransformer)
        +fit(dataset Dataset)
        +predict(X DataFrame) ndarray
    }

    class BaseTransformer {
        <<abstract>>
        +fit(X)*
        +transform(X)*
    }

    class BaseModel {
        <<abstract>>
        +fit(X, y)*
        +predict(X)*
    }

    class Dataset {
        +DataFrame X
        +Series y
        +split()
    }

    MLPipeline *-- BaseTransformer : composed of 0..*
    MLPipeline *-- BaseModel : composed of 1
    MLPipeline --> Dataset : uses
```
